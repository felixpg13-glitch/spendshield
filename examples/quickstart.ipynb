## Create the simplest shield

**Policy: maximum transaction = $50**

````python
from spendshield import SpendShield

shield = SpendShield(budget=100, max_amount=50, approve_new_recipient=False)
print("Shield ready — policy: max transaction = $50, daily budget = $100")
````

## Watch it work — $25 → ALLOW, $75 → DENY, $45 → ALLOW

Everything runs in this one cell:

````python
for amount, label in [(25, "within the $50 limit"), (75, "exceeds the $50 limit"), (45, "within the $50 limit")]:
    r = shield.authorize("", amount, "mcdonalds.com")
    icon = "✅" if r.decision == "ALLOW" else "❌"
    print(f"{icon} ${amount} → {r.decision}")
    print(f"   ${amount}.00 → mcdonalds.com — {label}")
````

## Why was $75 denied?

```
Transaction:        $75
Maximum allowed:    $50
Decision:           DENY
```

SpendShield doesn't simulate payments — it makes the **authorization decision before the money moves**.
