**Step 2:** Everything runs in the single cell below — watch SpendShield block a purchase:

```python
from spendshield import SpendShield

# Policy: max $50 per transaction, $100/day budget
shield = SpendShield(budget=100, max_amount=50, approve_new_recipient=False)

for amount, note in [(25, "within the $50 limit"), (75, "EXCEEDS the $50 limit"), (45, "within the $50 limit")]:
    r = shield.authorize("", amount, "mcdonalds.com")
    print(f"{'✅' if r.decision == 'ALLOW' else '❌'} ${amount} → {r.decision}   ({note})")
```

## You've just seen the core of SpendShield

- `$25 → ALLOW`  `$45 → ALLOW`  `$75 → DENY`

SpendShield doesn't simulate payments — it makes the **authorization decision before the money moves**.
