## Setup — one line

$100/day budget, $50 per-transaction cap:

````python
from spendshield import SpendShield

shield = SpendShield(budget=100, max_amount=50, dry_run=False,
                     approve_new_recipient=False)
````

## Scene 1 — Agent tries to spend $75

Policy says max $50 per transaction. Watch what happens:

````python
result = shield.authorize("", 75, "amazon.com")
print(f"{result.decision} — {result.reason}")
print(f"reason code: {result.rules[0].code if result.rules else '-'}")
````

## Scene 2 — Agent spends $25

Within limits → allowed:

````python
result = shield.authorize("", 25, "amazon.com")
print(f"{result.decision} — {result.reason}")
````

## Scene 3 — Splitting to dodge the budget

Three $40 payments against a $100 daily budget — the third one gets caught:

````python
for i in range(3):
    r = shield.authorize("", 40, "amazon.com")
    print(f"payment {i+1}: ${40} → {r.decision}")
    if r.decision == "APPROVAL":
        shield.approve(r.approval_id, by="you")
````
