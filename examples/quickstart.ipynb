## Create the simplest shield

**Policy: maximum transaction = $50**

````python
from spendshield import SpendShield

shield = SpendShield(budget=100, max_amount=50, approve_new_recipient=False)
````

## $25 purchase → within the $50 limit → allowed

````python
result = shield.authorize("", 25, "mcdonalds.com")
print(f"{'✅' if result.decision == 'ALLOW' else '❌'} {result.decision}")
print(f"$25.00 → mcdonalds.com")
print(f"Within the $50.00 transaction limit")
````

## $75 purchase → exceeds the $50 limit → **DENIED** ⭐

````python
result = shield.authorize("", 75, "mcdonalds.com")
print(f"{'✅' if result.decision == 'ALLOW' else '❌'} {result.decision}")
print(f"$75.00 → mcdonalds.com")
print(f"Transaction exceeds the $50.00 limit.")
````

## Why was it denied?

```
Transaction:        $75
Maximum allowed:    $50
Decision:           DENY
```

SpendShield doesn't simulate payments — it makes the **authorization decision before the money moves**.

## Try another amount — $45

````python
result = shield.authorize("", 45, "mcdonalds.com")
print(f"{'✅' if result.decision == 'ALLOW' else '❌'} {result.decision}")
print(f"$45.00 → mcdonalds.com")
````

## You've just seen the core of SpendShield
